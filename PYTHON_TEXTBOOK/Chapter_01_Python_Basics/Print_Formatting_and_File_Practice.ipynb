#   **printing using custom method**

```python

class Product:
    # Write your initialization component here:
    def __init__(self, item_name, price):
        self.item_name = item_name
        self.price = price
        
    # Write your custom __str__ method component below:
    def __str__(self):
        # Hint: Return a formatted string!
        return f"Product: {self.item_name} costs Rs. {self.price}"

# --- Test Environment ---
p1 = Product("Wireless Mouse", 1200)
print(p1) # Should print: Product: Wireless Mouse costs Rs. 1200
```
