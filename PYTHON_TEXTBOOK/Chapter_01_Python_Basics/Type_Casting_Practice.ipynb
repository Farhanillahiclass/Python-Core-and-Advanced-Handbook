# **2. Dynamic Type Casting**

# **Implicit Type Casting**


Implicit type casting (also called automatic type conversion or coercion) happens when the compiler or interpreter automatically converts one data type into another.
This occurs without any explicit instruction from the programmer.
It typically happens when you perform an operation involving two different data types.
To prevent data loss, the system automatically converts the "smaller" or less precise data type into the "larger" or more precise data type.

# **The Single Rule of Python Implicit Casting**
Unlike JavaScript, which aggressively coerces strings and numbers, Python is a strongly typed language. Python will only perform implicit casting between numeric types: integers, floats, and complex numbers.

The general rule is:
* _The type with higher precision always wins.int + int ➔ int_
* _int + float ➔ float (Integer is promoted to prevent losing the decimal)_

```python
x = 10    # Integer
y = 2.5   # Float
z = x + y  # x is implicitly cast to float (10.0)
print(z)   # Output: 12.5
```

# **Explicit Type Casting**

in Python, explicit type casting is the process of manually converting a variable from one data type to another using built-in functions.

Because Python is a dynamically typed language, it will not automatically convert data types if it risks losing information or context. You must explicitly tell Python to change the type by wrapping your variable or value inside a type constructor function.

##  **The Core Python Casting Functions**
**_Python provides four primary functions for explicit type casting:_**
* _int(): Converts data into an integer (whole number)._
* _float(): Converts data into a floating-point number (decimal)._
* _str(): Converts data into a string (text)._
* _bool(): Converts data into a boolean (True or False)._

```python
a = "0.90"
print(a)
print(type(a))
a = float(a)
print(type(a))

# output :
#0.90
#<class 'str'>
#<class 'float'>
```

# **Eval() in type casting**

In Python, eval() (short for evaluate) is a built-in function that takes a text string as an argument, parses it as a live Python expression, and executes it.
When used for type casting, eval() acts as an automatic, dynamic converter. 

Instead of you choosing a specific function like int(), float(), or list(), eval() inspects the contents of the string and instantly converts it into its matching, native Python data type.

# **The Rule for eval()**
 * _Contenteval("45") ➔ evaluates to number 45 ✅_
 * _eval("67.7") ➔ evaluates to number 67.7 ✅_
 * _eval("Ali") ➔ looks for a variable named Ali ❌_
 * _eval("'Ali'") ➔ evaluates to the text string 'Ali' ✅_

```python
#  for string we should put triple commas insted of quoted comma
a = '"Ali"'  # Nested quotes protect the text from being treated as a variable name
b = "45"
c = "67.7"

print(type(c))
print(a, b, c)

print("---" * 5)

a = eval(a)  # Strips outer quotes, leaves '"Ali"', casts to valid string object
b = eval(b)  # Strips outer quotes, leaves 45, casts to integer
c = eval(c)  # Strips outer quotes, leaves 67.7, casts to float

print(type(c))
print(a, b, c)
```

# **Changing Variable type Dynamically**

in  Python, dynamic typing allows you to change a variable's data type automatically simply by assigning a new value to it.Unlike statically typed languages (like Java or C++), you do not need to declare a variable's type beforehand.


 Python determines the type at runtime based on whatever value the variable holds at that exact moment.

```python
p= "ali"
print(type(p))
print(p)
print("____"*4)
p=10
print(type(p))
print(p)
print("____"*4)
p= 22.0987
print(type(p))
print(p)
```
