# **_Variables and its types_**
## Defination:
###  In Python, a variable is a named pointer or reference that points to a specific object or value stored in your computer's memory. Think of a variable as a labeled container or sticky note used to store, track, and manipulate data throughout your program.
## Key Rules for Naming Variables:
* ### Start correctly: Names must begin with a letter or an underscore (_).
* ### No leading numbers: Names cannot start with a digit (e.g., 1place is invalid).
* ### Allowed characters: Names can only contain alphanumeric characters and underscores (1. A-z, 0-9, and _).
* ### Case sensitivity: Names are case-sensitive (age, Age, and AGE are three separate variables).
* ### Avoid keywords: Reserved keywords (like if, else, for, or def) cannot be used.

```python
# examples :
a = "ali"
print(a)
print(type(a))
# here "a"   is a variable like a container and "ali" is a data which store in that container
```
