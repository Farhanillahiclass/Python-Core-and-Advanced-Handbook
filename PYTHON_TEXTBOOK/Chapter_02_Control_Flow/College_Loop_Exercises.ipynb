# **1 While Loop** 

In Python, a while loop is used to repeat a block of code continuously for as long as a specific condition remains True.

Think of a while loop like an automatic ticket barrier at a subway station. The barrier stays closed (True) and keeps blocking people while their card has no balance. The moment they swipe a valid card, the condition changes, and the loop stops.

# **Key Properties of a while Loop**
* _Indefinite Iteration: Unlike a for loop, a while loop does not know how many times it will run beforehand._ 
* _It only cares about the condition._
* _Requires a State Change: Inside the loop, you must change something that eventually makes the condition False._ 
* _If you do not, the loop will run forever._
* _The Infinite Loop Danger: If your condition never becomes False, your program will freeze or crash because it enters a never-ending loop._
* _Pre-test Nature: The condition is evaluated before the code runs._
* _If the condition starts as False, the code block inside will be completely skipped._

```python
a=int(input("Enter value"))
b=int(input("Enter value"))
while a<=b:
    print(a)
    a+=10
    
```

```python
c=int(input("Enter value"))
d=int(input("Enter value"))
while c<=d:
    print(c)
    c+=1
```

# **2 For Loop** 

### In Python, a for loop is used to repeat a block of code a specific number of times or to step through items in a collection (like a list, tuple, or string) one by one.  

_Think of a for loop like a factory conveyor belt. Items come down the belt one at a time, and a robotic arm performs the exact same action on every single item until the belt is empty._

# **Key Properties of a _for_ Loop** 
* _Definite Iteration: It knows exactly how many times it needs to run before it even starts._ 
* _It stops automatically when it reaches the end of the collection or sequence._
* _The Loop Variable: It uses a temporary variable (often named i, item, or x) that automatically updates to hold the value of the current item on each turn._
* _Requires Indentation: The block of code you want to repeat must be indented (shifted right by 4 spaces) under the for statement._
* _Works with range(): If you do not have a list but just want to repeat an action a set number of times (like 5 times), you pair the loop with Python's built-in range() function._
