# **Control Flow Statement**
Python executes code sequentially from top to bottom. Control flow statements alter this linear path by creating branches, loops, and jumps based on runtime conditions. Python relies strictly on whitespace indentation (typically 4 spaces) rather than curly braces {} or keywords like end to define execution blocks.

# Architectural Guide to Python Control Flow

## 1. Conditional Branching (`if`, `elif`, `else`)
Conditional statements evaluate expressions down to a Boolean value (`True` or `False`) to route program execution.

### Real-World Analogy: E-Commerce Checkout
An online shopping cart adjusts shipping fees dynamically based on the total cart value and user tier.

```python
# Real-World Script: Cart Calculator
cart_total = 120.0
is_premium_member = False

if cart_total >= 100.0 or is_premium_member:
    shipping_cost = 0.0
elif cart_total >= 50.0:
    shipping_cost = 5.0
else:
    shipping_cost = 10.0

print(f"Your shipping fee is: ${shipping_cost}")
```

### Deep Concept: Truthiness vs Falsiness
Python does not just evaluate explicitly boolean conditions. Every object has an inherent boolean truth value.
* **Falsy Values (Evaluate to `False`):** `None`, `False`, `0`, `0.0`, `""` (empty string), `[]` (empty list), `{}` (empty dict).
* **Truthy Values (Evaluate to `True`):** Any object with content, or non-zero numbers.

*Production Tip: Instead of writing `if len(my_list) == 0:`, write `if not my_list:` for clean, idiomatic Python.*

## 2. Iteration Structures (`for`, `while`)
Loops automate repetitive operations. Python splits iteration into definite collection loops and indefinite state loops.

### The `for` Loop: Collection Iteration
Used when you know the number of iterations in advance or are processing a sequence of items.

#### Real-World Analogy: Server Status Check
A DevOps monitoring script checks an array of cloud server addresses for performance metrics.

```python
# Real-World Script: Network Ping Test
servers = ["web-01", "database-01", "cache-01"]

for server in servers:
    print(f"Pinging {server}... Status: Connected")
```

### The `while` Loop: Condition-Based Iteration
Used when execution must continue indefinitely until a specific flag or state changes.

#### Real-World Analogy: Smart Home Thermostat
A heating system runs continuously until a physical temperature sensor detects the target room climate.

```python
# Real-World Script: Climate Engine
current_temp = 16.0
target_temp = 21.0

while current_temp < target_temp:
    print(f"Heater running. Current Temp: {current_temp}°C")
    current_temp += 0.5  # Modifying state to prevent an infinite loop

print("Target reached. Turning off heater.")
```

## 3. Loop Flow Modifiers (`break`, `continue`, `pass`, `else`)
Modifiers override normal loop behavior by jumping lines during runtime execution.

### `break` & `continue` Statements
* **`break`**: Instantly exits the entire loop.
* **`continue`**: Skips the remaining lines of code *for the current loop cycle* and forces the loop to evaluate the next item.

#### Real-World Analogy: Security Log Parser
Scanning a text file for unauthorized access alerts while filtering out safe internal traffic.

```python
# Real-World Script: Security Scanner
log_events = ["INFO: Login", "DEBUG: Memory Clear", "CRITICAL: Malicious IP", "INFO: Logout"]

for event in log_events:
    if "DEBUG" in event:
        continue  # Skip unneeded details and go to the next log item
    
    if "CRITICAL" in event:
        print(f"ALERT Triggered: {event}")
        break  # Lock system and immediately exit loop processing
```

### Unique Feature: The Loop `else` Block
An `else` statement attached to a loop executes **only if the loop finishes its cycle naturally** without hitting a `break` statement.

```python
# Real-World Script: Database Record Search
user_database = ["Alice", "Bob", "Charlie"]
search_target = "David"

for user in user_database:
    if user == search_target:
        print("User found!")
        break
else:
    # This executes because the loop scanned the whole list without a 'break'
    print("Error: User record does not exist in our system.")
```

## 4. Pattern Matching (`match-case`)
Introduced in Python 3.10, `match-case` structural pattern matching acts like a high-performance sorting machine for complex values, objects, and configurations.

### Real-World Analogy: Smart Home Command Processing
Processing voice command structures passing through an API network gateway.

```python
# Real-World Script: Smart IoT Controller
# Inputs could be standard strings or structured coordinate tuples
voice_command = ("light", "living_room", "on")

match voice_command:
    case "shutdown":
        print("Turning off entire household network.")
    case ("light", room, state):
        print(f"Setting the light in {room} to status: {state.upper()}")
    case ("temperature", setting):
        print(f"Adjusting thermostat to {setting} degrees.")
    case _:
        print("Voice command unrecognized. Please try again.")
```


# 📊 Summary Reference: Control Flow Guide

| Statement Keyword | Structural Purpose | Real-World Scenario |
| :--- | :--- | :--- |
| **`if`** | Run code only if a condition passes | Check if user entered a valid password |
| **`else`** | Run alternative code if the condition fails | Show an error message for an invalid password |
| **`elif`** | Add another condition check to a sequence ladder | Evaluate grades (A, B, C) based on test percentages |
| **`for`** | Iterate a predictable, fixed number of times | Scan through all 100 items inside a shopping cart list |
| **`while`** | Keep looping until a target event occurs | Keep an app open until the user clicks the "Exit" button |
| **`break`** | Kill the current loop immediately | Exit a file download loop early if an error is caught |
| **`continue`** | Skip the current step and proceed to the next cycle | Skip printing even numbers while looping through a list |
