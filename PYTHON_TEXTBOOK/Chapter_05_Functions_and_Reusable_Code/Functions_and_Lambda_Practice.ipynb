```python
def mean_of_list(numbers):
    """This function calculates the mean of a list of numbers"""
    # return the mean and round it off to 2 decimal places
    return round(sum(numbers) / len(numbers), 2)

price = [200, 300, 600, 550, 566, 1000, 2000, 3500, 2]
print(mean_of_list(price))  # Output: 968.67
```

```python
discount=[40,50,90,99,77,88,55,33]
print(mean_of_list(discount))
```

## **lambda function**

```python
# for displaying and understanding the lambda function ,firstly we should write a function in def and then a lambda function   with the same logic

# user_defined function 

def user_defined(a,b):
    """this is a user defined function which add two value
    """
    return a+b
    
```

```python
user_defined(90,90)
```

```python
# now same logic but in lambda 
add_lambda = lambda a,b:a+b
# Here add_lambda is a variable by which we call the function
# here  lambda a,b   is the function parameter 
''' here a+b calling the logic of function Its mean what will happen 
when user put data in lambda parameter
ITS working is similar to return funtion which used in user _defined function
such as def'''
```

```python
add_lambda(1,4)
```

```python
is_adult = lambda birth_year, current_year=2026: (current_year - birth_year) >= 18

```

```python
is_adult(2001)
```

```python
steps = [4000, 8500, 12000, 3100]

# Map aur lambda ka use
kilometers = list(map(lambda x: round(x / 1300, 2), steps))

print(kilometers)
# Output: [3.08, 6.54, 9.23, 2.38]
```

```python
items = [
    {"name": "Book", "price": 15},
    {"name": "Headphones", "price": 85},
    {"name": "Water Bottle", "price": 25},
    {"name": "Shoes", "price": 120}
]

# Filter aur lambda ka use
budget_items = list(filter(lambda item: item['price'] <= 50, items))

print(budget_items)
# Output: [{'name': 'Book', 'price': 15}, {'name': 'Water Bottle', 'price': 25}]
```
