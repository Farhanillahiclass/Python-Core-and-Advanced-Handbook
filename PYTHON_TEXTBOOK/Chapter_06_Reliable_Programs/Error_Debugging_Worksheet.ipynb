# **What is Error Or Bug ?**

# **Defination :**
_A programming bug is a mistake in a program's code that causes it to behave unexpectedly or fail._ 
           
***or***

_A programming error is the actual misstep, incorrect action, or flaw made by the programmer while writing the code_


# **Key Types of Bugs and** **Errors :** 
* _Broken grammar rules of the coding language.Runtime Errors: Crashes that happen while the program runs._
* _Logic Errors: Code runs perfectly but produces the wrong result.Compilation Errors: Mistakes that prevent code from converting into a runnable program._
* _Semantic Errors: Code that makes no sense to the computer's engine._

# **Debugging :**

_To find and resolve bugs in your code, you need a systematic approach called debugging._

# **How to Find Bugs**
* _Reproduce consistently: Find the exact steps that trigger the problem every single time._
* _Use print statements: Insert print() or console.log() to check variable values at different steps._
* _Run a debugger: Use your IDE's built-in debugger to pause code execution and inspect state._
* _Set breakpoints: Stop the program at specific lines to see exactly where it breaks._
* _Check error logs: Read the stack trace to find the exact file and line number._
* _Isolate code blocks: Comment out sections of code to find the problematic area._

# **How to Resolve Bugs**
* _Understand the root cause: Do not guess; ensure you know why the code failed before changing it._
* _Change one thing: Modify only one piece of code at a time to track what works._
* _Write unit tests: Create a test that reproduces the bug, then write code to pass it._
* _Explain it to someone: Use the "Rubber Duck" method by explaining your code aloud line-by-line._
* _Check code history: Use Git to see what recent changes might have introduced the bug._
* _Search for errors: Search the exact error message online on platforms like Stack Overflow._

# **How to Find and Fix Bugs Using AI**

Using AI to debug code is like having a 24/7 **teaching assistant.** However, AI can make **mistakes** if you give it bad instructions. To get the best results, you must **learn** how to prompt it correctly, verify its answers, and use it to learn—not just **copy-paste.**

# **1. The Anatomy of a Perfect AI Debugging Prompt**

AI needs context to give accurate fixes. When asking an AI for help, always structure your prompt with these five components:
* _The Technology: State the programming language, framework, or library version you are using._
* _The Goal: Clearly explain what the code should be doing._
* _The Problem: Explain what the code is actually doing (e.g., crashing, infinite loop, wrong output)._
* _The Evidence: Paste the exact error message or stack trace._
* _The Source: Paste the relevant block of code (keep it under 50 lines for best results)._

# **2. Fill-in-the-Blank Templates for Students**

# **Template A:** For Syntax or Runtime Errors (The Code Crashes)


```
"I am working on a assignment using [Language, e.g., Python 3.11]. My code is crashing with this exact error message: [Paste Error Message Here].My goal is to [State what the code should do]. Here is my code:[Paste Code Here]Can you explain what this error means in simple terms and show me how to fix it?"


```

# **Template B:** For Logic Errors (The Code Runs, But Gives the Wrong Output)


```
"I am writing a program in [Language]. The code runs without crashing, but it is giving the wrong result.Expected Output: [What it should say]Actual Output: [What it actually says]Here is my code:[Paste Code Here]Where is the logic flaw in my thinking, and how do I correct it?"
```

# **Template C:** For Edge Cases (Code Works for You, But Fails the Auto-Grader)


```
"This [Language] code works fine for my basic tests, but it fails when I submit it to my school's automated grading system.Can you review this code for hidden bugs, boundary conditions, or edge cases (like empty inputs, negative numbers, or large values) that I might have missed?[Paste Code Here]"
```

# 3. Golden Rules for Students (AI Etiquette & Safety)
* _Never Copy Blindly: Always ask the AI, "Why did this error happen, and how does your code fix it?" If you do not understand the fix, you cannot defend it in class or exams._
  
* _Debug Iteratively: If the AI gives you a fix and it causes a new error, do not panic.__
* _Paste the new error back into the AI and say, "That fix gave me this new error: [Paste Error]._
* _Isolate Your Code: Do not paste hundreds of lines of code._
* _Isolate the exact function or loop that is broken._ 
* _Large files confuse the AI and lead to messy answers._
* _Beware of Hallucinations: AI sometimes makes up functions, libraries, or syntax that do not exist._ 
* _If the code it provides looks completely unfamiliar, double-check your official class textbook or documentation._
* _Academic Integrity Check: Always check your syllabus._ 
* _Some professors allow AI for debugging but forbid it for writing original code.__ 
* _Never ask AI to "write my assignment from scratch."_
