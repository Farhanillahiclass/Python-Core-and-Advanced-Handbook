# **🏷️ Definitions:** 
## **Data Structures in Python**

A **Data Structure** is a specialized format or container used to organize, manage, and store data in computer memory so that operations can be performed efficiently.


# 🗺️ Data Structure Classification Flowchart

```text
Data Structure
 ├── Primitive
 │    ├── Int
 │    ├── Char
 │    ├── Bool
 │    ├── Float
 │    └── Pointer
 │
 └── Non-Primitive
      ├── Linear
      │    ├── Static
      │    │    └── Array
      │    │
      │    └── Dynamic
      │         ├── Linked list
      │         ├── Stack
      │         └── Queue
      │
      └── Non-linear
           ├── Tree
           └── Graph
```


# **Complete Classification of Data Structures**

This guide breaks down data structures exactly as shown in the structural hierarchy map, dividing them into **Primitive** and **Non-Primitive** structures.

---

# PART 1: Primitive Data Structures

Primitive data structures are the most basic, fundamental data blocks provided natively by a programming language. They store a single, simple value at a time and cannot be broken down into smaller structures.

## 1. Integer (`Int`)
*   **When to use:** Use to store absolute, whole numbers without any fractional or decimal values.
*   **Examples:** Counting total loops, age values, or user ID sequence numbers.

## 2. Character (`Char`)
*   **When to use:** Use to store a single textual alphabet letter, single number digit, punctuation symbol, or space indicator.
*   **Examples:** Storing grading scales (`'A'`, `'B'`), user gender inputs (`'M'`, `'F'`), or individual keystrokes.

## 3. Boolean (`Bool`)
*   **When to use:** Use to handle absolute true or false states for conditional checks and logic flags.
*   **Examples:** System status states like `True` for logged-in or `False` for system errors.

## 4. Float (`Float`)
*   **When to use:** Use to represent real, continuous numerical values that require precise decimal points.
*   **Examples:** Storing precise scientific parameters, financial item costs, or physical body weights.

## 5. Pointer (`Pointer`)
*   **When to use:** Use to explicitly hold the direct hex memory address location of another variable instead of storing a raw data value.
*   **Examples:** Managing system resources, navigating nested array addresses, or linking memory dynamically.

# PART 2: Non-Primitive Data Structures

Non-Primitive data structures are complex containers built by grouping primitive variables together. They organize multiple values systematically inside memory.

---

## 🛑 Sub-Branch A: Linear Data Structures
In Linear structures, elements are chained together sequentially in a straight row. Every item connects directly to the one right before and right after it.

### 📍 Category 1: Static Linear Structures
Static structures have a fixed size locked inside computer memory. Once declared, their capacity cannot grow or shrink while the program is running.

#### **Array**
*   **The Analogy:** A row of locked gym lockers numbers 1 to 10 built into a concrete wall. You cannot add locker #11 without tearing down the wall.
*   **When to use:** Use when you know the exact maximum count of items ahead of time and need instant positional tracking.
*   **Examples:** Storing a fixed matrix of 12 monthly sales values or defining a permanent 7-day calendar index.
   []

---

### 📍 Category 2: Dynamic Linear Structures
Dynamic structures have an adjustable, flexible size. They scale automatically, expanding or shrinking in memory as you add or remove elements.

#### **Linked List**
*   **The Analogy:** A children's paperclip chain. You can unhook any two paperclips in the middle to slide a new one in smoothly without altering the rest.
*   **When to use:** Use when your collection size changes constantly and you need to insert or delete elements in the middle quickly.
*   **Examples:** Managing audio tracks inside a music player playlist queue.
  

#### **Stack (LIFO - Last In, First Out)**
*   **The Analogy:** A vertical stack of clean dinner plates. The last plate placed on top is always the very first plate lifted away.
*   **When to use:** Use when you need to track steps sequentially and reverse your actions backward.
*   **Examples:** Storing history logs for an "Undo" command (`Ctrl + Z`) or tracking your browser history steps.

#### **Queue (FIFO - First In, First Out)**
*   **The Analogy:** A single-file line of people waiting at a bus stop. The first person to arrive stands at the front and gets on the bus first.
*   **When to use:** Use when processing items or tasks chronologically in the exact order they arrive.
*   **Examples:** A computer printer buffer managing multiple incoming print requests from an office network.

---

## 🛑 Sub-Branch B: Non-Linear Data Structures
In Non-Linear structures, elements are not arranged in a straight row. Instead, items are interconnected hierarchically or in complex web structures across memory.

#### **Tree**
*   **The Analogy:** A family tree. Ancestors sit at the top ("Root"), branching down to parent generations, who branch down further to child generations.
*   **When to use:** Use when your data follows a strict top-down, parent-child, or ancestral ownership structure.
*   **Examples:** Your computer operating system's internal folder layout (`C: Drive ➔ Documents ➔ Homework ➔ project.py`).
    

#### **Graph**
*   **The Analogy:** A global airline flight destination map. Multiple destination cities are completely interconnected by flight pathways traveling back and forth.
*   **When to use:** Use when data elements have multi-directional, networked relationships without a clean top or bottom.
*   **Examples:** Social media friend matrices (mapping connection links on LinkedIn or Facebook) or city navigation grids inside map routing software.
