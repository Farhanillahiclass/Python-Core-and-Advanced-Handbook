```python
def fibonacci_while(max_value):
    a, b = 0, 1
    while a <= max_value:
        print(a, end=" ")
        a, b = b, a + b

# Run the function
fibonacci_while(1000)
```

# **pass in python**

Aisi soorat mein hum pass likh dete hain. pass ka matlab hota hai: "Python bhai, abhi yahan kuch nahi hai, bina error diye aage guzar jao (pass ho jao)."

```python
def add_numbers(num1, num2):
    # 'print' karne ki jagah humne 'return' likha
    return num1 + num2  

# Ab humne function chalaya aur uska answer 'result' naam ke variable mein save kar liya
result = add_numbers(10, 20)

# Ab is 'result' ko hum aage kisi bhi kaam mein use kar sakte hain
print(result * 2)  # Output: 60 (30 ko 2 se multiply kiya)
```
