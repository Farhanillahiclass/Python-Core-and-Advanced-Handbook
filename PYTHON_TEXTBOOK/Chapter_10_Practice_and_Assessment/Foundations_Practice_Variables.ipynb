## **2 : Integer**

```python
# in python integer (int) is a number or whole number ,it may be positive or negative
age = 12
total_marks = 2000
gain_marks = 1899
cutting_marks = -101
result_percentage = 88 # %
print(age)
print(total_marks)
print(gain_marks)
print(cutting_marks)
print(result_percentage)
print(type(age))  # <class 'int'>
```

## **3 : Float**

```python
# In Python, a float (floating-point number) is simply a number with a decimal point. It is the data type Python uses to represent fractional or precise values instead of whole numbers.
a = 10.4
jupiter_distance_au = 5.20
mars_distance_km = 227900000.0
punjab_services = 0.16  # Punjab Revenue Authority (16%)
item_price = 1000.0
gst_rate = 0.18
print(a)
print(jupiter_distance_au)
print(mars_distance_km)
print(punjab_services)
print(item_price)
print(gst_rate)
print(type(a))  # output : <class 'float'>
```

## **4 : Boolean**

```python
# In Python, a boolean (bool) is a data type that has only two possible values: True or False. It acts like a digital light switch that is either ON or OFF to help your program make decisions.
a = True
payment_status = True
car_working = False
global_lockdown = False
privacy_policy = True
print(a)
print(payment_status)
print(car_working)
print(global_lockdown)
print(privacy_policy)
print(type(a)) # output : <class 'bool'>
```

# **7 : Dictionary**

```python
# In Python, a dictionary (dict) is simply a collection of key-value pairs stored inside curly brackets {}. It acts exactly like a real-world dictionary or a phone book: you look up a specific word (the key) to find its definition or number (the value).
roll_no_01 ={"name":"Muhammad Farhan","age":16,"Subject":"Artificial Intellligence","level":"Beginner"}
roll_no_02 ={"name":"Muhammad Rehan","age":12,"Subject":"Artificial Intellligence","level":"Beginner"}
roll_no_03 ={"name":"Muhammad Anas","age":17,"Subject":"Video Editing","level":"Beginner"}
agent_00y ={"name":"Annonyamus","age":50,"Subject":"Acting","level":"Advance"}
print(roll_no_01["Subject"])
print(roll_no_02["name"])
print(roll_no_03["level"])
print(agent_00y["age"])

#  Modifying a value
roll_no_01["Subject"] = "Researcher and ai Ai Engineer"   # Changes the Subject to Researcher and ai Ai Engineer

#  Adding a new key-value pair
agent_00y["role"]= "senior"  # Adds a new entry into the dictionary
# checking that the new entry is added
print(agent_00y["role"])
#  Checking the data type
print(type(roll_no_01))        # Outputs: <class 'dict'>
```

# **8 : SET**

```python
# In Python, a set is an unordered collection of unique items stored inside curly brackets {}. It acts like a digital sorting basket that automatically filters out duplicates, ensuring that every single item inside appears exactly once.
#Think of it like a bag of lottery balls. You can throw items into the bag, but you cannot have two balls with the exact same number, and they do not stay in any specific row or order.

my_skills ={"python programming","English basics","Master of Markdown","English basics",}   # here see the duplicate skill such as English  basics   ,it will automatically remove by python when you you print and show only single
simple_set ={2,4,8,7,4,9,8,67,12}
print(my_skills)
print(simple_set)
# add more item in variable 
my_skills.add("Master of Django files handling")
# remove any item
my_skills.remove("English basics")
print(my_skills)
print(type(simple_set))  # output : <class 'set'>
```

# **9 : None** 

```python
# In Python, None is a special keyword used to define a null value or no value at all.Think of it like an empty box. The box exists, but there is absolutely nothing inside it. It does not mean 0 (which is a number), and it does not mean "" (which is an empty string). It simply means "nothing."

rehman_portfolio_update = None
print(rehman_portfolio_update)
print(type(rehman_portfolio_update))
# output : <class 'None'>

# None is only when there is nothing in variable and when you add any databof any format in it  ,it data type will change according to new data format such as int,bool,set,tuple,str .....etc
rehman_portfolio_update = str("yes")
print(rehman_portfolio_update)
print(type(rehman_portfolio_update))
# output:<class 'str'>

```

# **10 : Complex**

```python
# In Python, complex is a built-in numeric data type used to represent complex numbers.Think of it as a number with two separate parts bound together: a real part that you use in everyday math, and an imaginary part used for advanced calculations.In mathematics, imaginary numbers use an i, but in Python, the letter j is strictly used.


num_01 = 7+8j   # directly putting the complex value
print(num_01)
num_02 = complex(99-90) # making complex number using complex function
print(num_02)
print(type(num_02))
# output : <class 'complex'>
```

# **3. Type casting**

```python
# Type casting (also known as type conversion) is the process of changing a variable from one data type to another.
# 
# Think of it like melting a piece of ice into water. The core substance remains the same, but the shape and format change so it can fit into a different container or be used for a different purpose.

num_01 ="0.01"
print(type(num_01))  # output : <class 'str'>
num_02 =float(num_01)
print(type(num_02))  # after prototyping  , output : <class 'float'>
string_01= "1,2,2,5"
print(type(string_01))  # output : <class 'str'>
string_02 = set(string_01)
print(type(string_02))   # after prototyping  , output : <class 'set'>
```
