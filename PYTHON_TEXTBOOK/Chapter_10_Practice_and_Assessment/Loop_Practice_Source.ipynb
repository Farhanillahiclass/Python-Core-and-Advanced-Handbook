# **Loops In Python**
By default, a computer executes code sequentially from top to bottom.

Loops are control flow structures that break this straight line. They tell Python to repeat a specific block of code multiple times automatically.Instead of copying and pasting the same line of code 100 times, a loop allows you to write it once and tell the computer to rerun it under a specified condition. 
 
 Think of a loop like a running track: the athlete keeps running around the oval loop until they complete their target number of laps or get too tired to continue.

# 🔁 Understanding Loops in Python

In programming, running the same block of instructions repeatedly is called **Iteration**. Python provides two primary looping mechanisms to handle iteration: **`for` loops** and **`while` loops**.

---

## 1. The `for` Loop (Definite Iteration)

*   **When to use:** Use a `for` loop when you know **exactly how many times** you want to repeat ahead of time, or when you want to scan through elements inside a sequence container (like a List, Tuple, String, or Range).
*   **The Mechanics:** Python automatically picks the next item in the sequence, assigns it to a temporary variable, runs the internal code block, and moves forward until the sequence runs out of items.

### Code Example (Looping through a List):
```python
fruits = ["Apple", "Banana", "Cherry"]

for fruit in fruits:
    print(f"I want to eat an {fruit}")
```

### Code Example (Looping using `range()`):
```python
# Rerun a script block exactly 3 times
for index in range(3):
    print(f"Loop count: {index}")  # Outputs index values: 0, 1, 2
```

---

## 2. The `while` Loop (Indefinite Iteration)

*   **When to use:** Use a `while` loop when you **do not know the exact repetition count** ahead of time. The loop acts like a continuous `if` statement—it will keep repeating your code block indefinitely as long as its core condition remains **`True`**.
*   **🚨 The Infinite Loop Hazard:** If the condition never changes to `False`, the loop will run forever, crashing your browser or laptop. You must write an **increment or update statement** inside the block to eventually break the loop.

### Code Example:
```python
countdown = 3

while countdown > 0:
    print(countdown)
    countdown -= 1  # 💡 Crucial: Changes the condition value to prevent a permanent lock!

print("Blast off! 🚀")
```

---

## 3. Loop Control Statements (Changing Paths on the Fly)

Python features three special keywords that let you manipulate or jump out of active loops instantly based on internal conditions:

### A. The `break` Keyword
*   Instantly terminates the loop completely, skipping any remaining steps and jumping straight to the rest of the script.
*   ```python
    for num in range(1, 6):
        if num == 3:
            break  # Exits the entire loop early when it hits 3
        print(num) # Outputs: 1, 2
    ```

### B. The `continue` Keyword
*   Skips *only* the current step's remaining lines and leaps straight back to the very top of the loop to process the next cycle.
*   ```python
    for num in range(1, 5):
        if num == 3:
            continue  # Skips printing 3 and jumps straight to number 4
        print(num)    # Outputs: 1, 2, 4
    ```

### C. The `pass` Keyword
*   A null statement placeholder. It does absolutely nothing, but it prevents syntax crashes when you want to leave a loop block empty to fill in later.

---

## 🛑 4. The Indentation Trap in Loops

Just like conditional statements, Python relies completely on **whitespace indentation** to determine which lines of code belong *inside* the loop and which lines sit *outside*.

```python
for i in range(2):
    print("This line is INSIDE the loop and runs twice.")
print("This line is OUTSIDE the loop and runs only ONCE at the very end.")
```


# 📊 Quick Reference: Loops Selection Guide

| Loop Type / Keyword | Core Operational Rule | Best Real-World Scenario |
| :--- | :--- | :--- |
| **`for` loop** | Iterates over a fixed, predictable list or range. | Sending a custom email greeting to all 50 users on a subscription list. |
| **`while` loop** | Repeats continuously until a true state flips to false. | Keeping a video game running continuously until the user clicks the "Quit" button. |
| **`break`** | Destroys the loop instantly. | Stopping a heavy file search loop the exact microsecond a match is successfully found. |
| **`continue`** | Skips the current round and restarts the next lap. | Scanning a customer log sheet but skipping over deactivated profiles without printing them. |


# **while Loops in Python**

A while loop is used for Indefinite Iteration, meaning it repeats a block of code an unpredictable number of times. Instead of stepping through a fixed sequence like a for loop, a while loop behaves like a continuous if statement. It evaluates a condition before every cycle; if the condition is True, it runs the code block. It will keep looping indefinitely until that condition suddenly becomes False.

Think of it like an automated lawn sprinkler: it keeps spraying water while the timer sensor detects that the soil is dry. The moment the soil becomes wet (Condition = False), the sprinkler turns off.

# 🔁 **Understanding `while` Loops in Python**

A **`while` loop** repeatedly executes a target block of statements as long as its controlling conditional expression evaluates to `True`. 

---

## 1. Core Syntax Anatomy

The standard layout of a Python `while` loop requires three distinct parts to work safely:

```text
1. INITIALIZATION  ➔   count = 1
2. LOOP CONDITION  ➔   while count <= 3:
                           print(count)
3. UPDATE STEP     ➔       count += 1
```

*   **1. Initialization:** You must create and set a starting value for your tracking variable *before* the loop starts.
*   **2. Loop Condition:** A boolean expression checked at the start of every lap. If it is `False` right at the beginning, the loop never runs.
*   **3. Update Step (Crucial):** A statement inside the block that modifies the tracking variable so the condition eventually becomes `False`.

---

## 2. 🚨 The Fatal Hazard: Infinite Loops

If you forget to write the **Update Step**, or if you write a condition that can never become false, you create an **Infinite Loop**. The computer will execute the block forever, locking up your notebook cell, consuming 100% of your CPU power, and potentially crashing your editor.

```python
# ❌ DANGEROUS INFINITE LOOP EXAMPLE (Do not run this!):
timer = 5
while timer > 0:
    print("Running forever...")
    # Bug! Forgot to subtract 1 from timer, so 5 is ALWAYS greater than 0.
```

### How to Force Stop a Stuck Notebook Cell:
If you or a student accidentally triggers an infinite loop, click the square **Interrupt Kernel** button (⏹️) at the top of your Jupyter Notebook toolbar to kill the process.

---

## 3. Real-World `while` Loop Design Patterns

Because `while` loops don't need a fixed range size, they excel at handling user inputs and system states where the end point is completely unpredictable.

### Pattern A: The User Input Sentinel (Validation Gate)
This loop forces a user to keep repeating their entry until they provide a valid response.
```python
password = ""

while password != "secret123":
    password = input("Enter your master password: ")
    if password != "secret123":
        print("Incorrect! Please try again.")

print("Access Granted!")
```

### Pattern B: The Game Loop Flag (`while True`)
Many systems use an intentional infinite loop combined with a `break` statement to run a continuous background app cycle.
```python
while True:
    command = input("Type 'play' to continue or 'exit' to quit: ").lower()
    if command == "exit":
        print("Closing application...")
        break  # 🎯 Instantly kills the permanent loop
    print("Game is running...")
```

---

## 4. The Advanced `while-else` Structure
Just like `for` loops, you can attach an **`else`** block to a `while` loop. 
*   The `else` block executes *only if* the loop ends naturally because the main condition turned `False`.
*   If the loop is killed early using a `break` statement, the `else` block is completely skipped.
```python
energy = 3
while energy > 0:
    print("Working...")
    energy -= 1
else:
    print("System Notification: Battery empty, loop ended naturally.")
```


# **for Loops in Python**
In programming, running a block of instructions repeatedly is called Iteration. A for loop is used for Definite Iteration, meaning it repeats a block of code a predictable number of times.
Instead of checking a condition like a while loop, a for loop stepping through an ordered sequence (like a List, Tuple, String, or Range). It automatically grabs the first item, runs the internal code, moves to the next item, and gracefully stops on its own when it runs out of data elements.

# 🔁 **Understanding `for` Loops in Python**

A **`for` loop** acts as an automated conveyor belt. It pulls individual items out of a collection container one by one, tracks the item inside a temporary variable, and executes your code block until the container is empty.

---

## 1. Core Syntax Anatomy

The standard layout of a Python `for` loop looks like this:

```text
               for item in sequence:
                    │         │
         ┌──────────┘         └──────────┐
         ▼                               ▼
  ITERATION VARIABLE             TARGET SEQUENCE
(Holds current element)       (List, String, or Range)
```
*   **The Colon (`:`):** Marks the entrance gate to the loop's execution block.
*   **The Indentation:** Every line shifted forward by 4 spaces runs repeatedly during each cycle of the loop.

---

## 2. Common Iteration Scenarios

### A. Stepping Through a List or Tuple
The loop automatically binds the iteration variable to the actual values inside the array, moving from index `0` straight to the end.
```python
students = ["Ali", "Sara", "Zain"]

for name in students:
    print(f"Good morning, {name}!")
# Outputs:
# Good morning, Ali!
# Good morning, Sara!
# Good morning, Zain!
```

### B. Iterating Over a Text String (`str`)
Because a string is an ordered sequence of individual characters, a `for` loop will step through it letter-by-letter, including blank spaces.
```python
word = "Code"
for letter in word:
    print(letter.upper())  # Outputs: C, O, D, E on separate lines
```

### C. Running a Loop N Times via `range()`
When you do not have an existing list but need code to repeat an exact number of times, combine the `for` loop with the `range(start, stop, step)` sequence tool.
```python
# Print a message exactly 3 times (Indexes: 0, 1, 2)
for i in range(3):
    print(f"Loading data slice... (Attempt {i})")
```

---

## 3. Advanced Technique: The `for-else` Block

Python features a unique structural logic chain: you can attach an **`else`** block directly beneath a `for` loop. 
*   The `else` code block executes automatically **only if the loop finishes its entire path naturally** without hitting a structural `break` statement.
*   If the loop is killed early by a `break` keyword, the `else` block is completely skipped. This is perfect for searching through databases.

```python
inventory = ["pen", "pencil", "eraser"]
search_target = "notebook"

for item in inventory:
    if item == search_target:
        print("Found the item in stock!")
        break
else:
    print("Search complete: Item is out of stock.")  # 🎯 Triggers because loop finished safely
```

---

```python
# ❌ LOGIC BUG:
for i in range(1, 4):
    print(i)
    i = 10  # ⚠️ This step is completely useless! Python resets 'i' to the next step on the next loop lap.
```

Never add (`.append()`) or remove (`.remove()`) items from a list while you are actively looping through it. Doing so alters the index references dynamically, causing Python to skip items or trap you in an infinite loop.
```python
numbers = 
for n in numbers:
    if n == 2:
        numbers.remove(n)  # ❌ DANGEROUS! Shifts the remaining indices silently.
```

# **break and continue in Python** 
In Python, loops normally run until their condition becomes false or they run out of items. However, sometimes you need to interrupt this normal path. break and continue are loop control transfer statements that allow you to change the behavior of an active loop on the fly based on internal conditions.
* Think of a loop like a car driving on a racetrack laps:break is like crashing through the exit gate: It stops the car instantly, cancels the entire race, and drives away from the track completely.

* continue is like skipping the rest of the current lap: The driver stops their current lap halfway through, teleports straight back to the starting line, and immediately begins the next lap.

# 🔄 **Loop Control Using `break` and `continue`**

Loop control statements allow you to alter the execution flow of `for` and `while` loops dynamically based on specific conditions evaluated *inside* the loop block.

---

## 1. The `break` Statement (The Emergency Exit)

*   **What it does:** Instantly terminates the loop completely. Python stops running any lines below the `break` keyword, exits the loop structure entirely, and moves on to execute the rest of the script below the loop.
*   **When to use:** Use `break` when you are searching for something and find it early, catching a fatal error, or intentionally exiting an infinite loop (`while True`).

### Code Example (`for` loop search optimization):
```python
names = ["Ali", "Sara", "Zain", "Hamza"]

for user in names:
    if user == "Zain":
        print("Target located! Stopping the search.")
        break  # 🎯 Exits the loop entirely. It will never check "Hamza".
    print(f"Checking user: {user}")

print("Loop completely finished.")
# Outputs:
# Checking user: Ali
# Checking user: Sara
# Target located! Stopping the search.
# Loop completely finished.
```

---

## 2. The `continue` Statement (The Lap Skipper)

*   **What it does:** Skips *only* the remaining code lines inside the current loop cycle. It forces Python to jump straight back to the top of the loop to evaluate the condition or pull the next sequence item for the next lap.
*   **When to use:** Use `continue` when you want to filter out or ignore specific data items (like empty files or bad inputs) without killing the entire loop structure.

### Code Example (Filtering data):
```python
for num in range(1, 6):
    if num == 3:
        print("Found 3! Skipping the print line below.")
        continue  # 🔄 Jumps back to the top to grab number 4
    print(f"Processing number: {num}")

# Outputs:
# Processing number: 1
# Processing number: 2
# Found 3! Skipping the print line below.
# Processing number: 4
# Processing number: 5
```

---

## 🛑 3. Crucial Pitfalls and Tricky Exam Traps

### Trap #1: The `while` Loop Infinite Continue Trap
If you use `continue` inside a `while` loop, you must ensure that your **Update Step (increment/decrement)** occurs *before* the `continue` statement runs. If the update step sits beneath `continue`, Python will skip it, the tracking variable will never change, and your loop will be trapped in an infinite loop!

```python
# ❌ THE INFINITE LOOP BUG:
count = 0
while count < 4:
    count += 1
    if count == 2:
        continue  # Bug! It jumps to the top, count stays 2 forever, freezing the app.
    print(count)
    count += 1

# ✅ THE FIXED WAY:
count = 0
while count < 4:
    count += 1  # Increment occurs BEFORE the check
    if count == 2:
        continue
    print(count)
```

### Trap #2: Nesting Boundaries (Only Affects the Immediate Loop)
If you place a `break` or `continue` statement inside a **nested loop** (a loop inside another loop), it **only** controls the specific inner loop it lives in. It cannot jump out of or control the outer loop.

```python
for outer in ["Row1", "Row2"]:
    for inner in:
        if inner == 2:
            break  # ⚠️ This ONLY kills the inner loop. The outer loop keeps running!
        print(f"{outer} - Slot {inner}")
```

# **_pass_ Statement in Python**

By default, Python cannot handle an empty block of code. If you create a conditional statement (if/else) or a loop (for/while) and leave its body completely blank, Python will crash instantly with a fatal IndentationError.

The pass statement is a special null or empty statement in Python. It acts as a structural placeholder that does absolutely nothing when executed. It simply tells Python: "I know a code block belongs here, but I haven't written it yet—please skip this block and move on without crashing.

"Think of pass like a "Coming Soon" sign at a construction site. The sign takes up physical space so the structure looks complete, but no actual building work is happening inside yet.

# 🗂️ **Understanding the `pass` Statement in Python**

The **`pass`** statement is a syntactic placeholder in Python. It is used when a statement is syntactically required by the language interpreter, but you want no code or action to execute at that moment.

---

## 1. Why `pass` is Necessary (The Python Indentation Rule)

In other languages like Java or C++, you can create an empty placeholder block using empty curly braces `{ }`. However, because Python relies entirely on **whitespace indentation** to group code, an empty block is a syntax violation.

### The Crash Example:
```python
# ❌ THIS CRASHES INSTANTLY:
if score >= 90:
    # Intending to write premium user logic here later...
else:
    print("Standard user.")
# Error Output: IndentationError: expected an indented block after 'if' statement
```

### The Correct Placeholder Example:
```python
# ✅ THIS RUNS PERFECTLY WITHOUT CRASHING:
if score >= 90:
    pass  # Placeholder: Code structure is satisfied, Python safely moves to 'else'
else:
    print("Standard user.")
```

---

## 2. Common Real-World Use Cases for `pass`

### Case A: Sketching Code Architecture (Stubbing)
When you are designing a large program, it is common practice to map out your loop structures, conditional paths, or functions ahead of time before actually writing the inner math logic. This is called **Code Stubbing**.
```python
def calculate_tax(income):
    pass  # To be coded by the finance team later

def process_payment():
    pass  # To be coded by the backend team later
```

### Case B: Creating Minimal Custom Exceptions
In advanced object-oriented programming, if you want to create a unique, custom error flag for your application, you can inherit from Python's base Exception class and use `pass` since you don't need to add new properties.
```python
class AccountLockedError(Exception):
    pass  # Inherits all basic error features without adding messy code
```

### Case C: Intentionally Ignoring Filtered Items in Loops
Sometimes inside a loop, you want a specific condition to explicitly do nothing, while other items drop down to be processed.
```python
for num in range(1, 6):
    if num % 2 == 0:
        pass  # Do nothing for even numbers
    else:
        print(f"Odd number found: {num}")
```

---

## 🛑 3. The Big Trap: `pass` vs. `continue` vs. `break`

Students frequently confuse `pass` with other loop control keywords. While they look similar, their impact on an active loop is completely opposite.

*   **`pass` (The Placeholder):** Does absolutely nothing. Python executes it and immediately moves down to run the very next line directly below it *within the same cycle*.
*   **`continue` (The Lap Skipper):** Abruptly stops the current cycle and forces Python to leap straight back to the top of the loop for the next round.
*   **`break` (The Emergency Exit):** Destroys the loop completely and exits to the rest of the script.

### Visual Comparison Check:
```python
# Scenario: What happens when num == 2?

for num in:
    if num == 2:
        pass
    print(num)  # 🎯 Outputs: 1, 2, 3 (2 is still printed because pass did nothing!)

for num in:
    if num == 2:
        continue
    print(num)  # 🎯 Outputs: 1, 3 (2 is skipped because continue jumped to the top!)
```

# **Infinite loops in python**
An infinite loop occurs when a loop keeps repeating indefinitely because its controlling condition never turns False

Think of it like an automated alarm clock that keeps ringing because the stop button is broken—it will keep making noise forever until you pull the battery out.

# ♾️ **Understanding Infinite Loops in Python**

An **Infinite Loop** is an execution path that repeats a block of code continuously without ever hitting a natural termination point. While most infinite loops are caused by unintentional programming bugs, developers sometimes create them on purpose to run continuous processes (like game engine cycles or background server monitors).

---

## 1. How Unintentional Infinite Loops Happen

Most infinite loops occur inside **`while` loops** because of three common mistakes:

### Mistake A: Forgetting the Update Step (Increment/Decrement)
If you do not change the tracking variable inside the loop body, the condition that was true at the start remains true forever.
```python
# ❌ THE UNINTENTIONAL BUG (Do not run this!):
count = 1
while count <= 5:
    print(f"Looping... Count is {count}")
    # Bug! Forgot to add 'count += 1', so 'count' is ALWAYS 1, which is always <= 5.
```

### Mistake B: Moving in the Wrong Direction
If your update statement moves your variable away from the stopping boundary instead of toward it, the loop can never finish.
```python
# ❌ THE DIRECTIONAL BUG (Do not run this!):
timer = 10
while timer > 0:
    print(timer)
    timer += 1  # Bug! Adding 1 makes the timer 11, 12, 13... It will never drop below 0!
```

### Mistake C: Float Imprecision and Strict Equality (`==`)
Computer processors sometimes store decimal numbers with tiny trailing errors. If you use strict equality (`==`) on a float to stop a loop, it might miss the exact stop value entirely.
```python
# ❌ THE FLOAT ACCURACY BUG:
x = 0.0
while x != 1.0:
    print(x)
    x += 0.1  # Due to internal binary math, x might jump from 0.99999 to 1.1, missing 1.0 completely!
```

---

## 2. Intentional Infinite Loops (`while True`)

Programmers often use a deliberate infinite loop pattern when they do not know when a task will finish, combined with a **`break` statement** to exit safely once a specific event occurs.

### Example: Continuous User Menu System
```python
while True:
    choice = input("Enter '1' for balance, '2' for settings, or 'q' to Quit: ").lower()
    
    if choice == 'q':
        print("Exiting application safely...")
        break  # 🎯 The terminal exit gate that kills the intentional infinite loop.
        
    print(f"Processing option {choice}...")
```

---

## ⏹️ 3. Emergency Protocol: How to Kill a Stuck Loop

*   This sends a `KeyboardInterrupt` signal to Python, forcing the infinite loop to crash safely so you can fix your code.
