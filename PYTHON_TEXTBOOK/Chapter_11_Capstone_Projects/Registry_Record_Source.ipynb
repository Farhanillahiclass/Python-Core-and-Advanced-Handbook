# **1.  String**

In Python, a string is simply text wrapped inside quotation marks. Technically, it is an ordered sequence of characters (like letters, numbers, spaces, or symbols) that Python treats as a single piece of data.

# Key Properties of a String
* _Enclosed in Quotes: You must wrap strings in either single quotes 'text' or double quotes "text"._ 
* _They work exactly the same way.Immutable: Once you create a string, you cannot change individual characters inside it._ 
* _If you modify a string, Python actually creates a brand-new string behind the scenes.Ordered and Indexed: Like lists, every character in a string has a position number starting at 0._
* _Multi-line Support: You can create long paragraphs that span multiple lines by wrapping the text in triple quotes ("""text""" or '''text''')._

````python
# Examples of string(str) vaariable

name = "Alice"          # Double quotes
city = 'London'         # Single quotes
message = """Hello,
this is a long text.""" # Triple quotes for multiple lines
print(name)
print(city)
print(message)

a = "Ali"
print(type(a))
# output :
# Alice
# London
#Hello,
#this is a long text
# <class 'str'>
````

# **3. Float**

 In Python, a float (floating-point number) is simply a number with a decimal point. It is the data type Python uses to represent fractional or precise values instead of whole numbers.

 on otherhand,
 In Python, a float (short for "floating-point number") is a data type used to represent real numbers that contain a decimal point.
 
 
 Think of it like measuring things precisely. While you use integers to count whole objects (like 3 cars), you use floats to measure continuous values that require accuracy, like 3.5 kilometers or a temperature of 98.6 degrees.

# Key Properties of a Float 
* _The Presence of a Decimal Point: A number is classified as a float the moment it includes a decimal point, even if the value after the decimal is zero (e.g., 5 is an integer, but 5.0 is a float)._
* _Scientific Notation Support: For extremely large or tiny numbers, floats use the letter e or E to represent powers of 10._
```
light_speed_approx = 3e8  # Means 3 * 10^8, stored as 300000000.0
```

* _The Result of Division: Any time you divide numbers using a single slash /, Python automatically outputs a float, regardless of whether the numbers divide perfectly._
```
print(4 / 2)  # Outputs: 2.0
```
* _Floating-Point Imprecision: Computer hardware stores floats in binary format (base 2), which cannot perfectly represent certain base-10 decimals. This sometimes causes tiny calculation anomalies._
```
print(0.1 + 0.2)  # Outputs: 0.30000000000000004 (instead of exactly 0.3)
```

````python
# Example of float(float) vaariable

price = 19.99      # A price
pi = 3.14159       # A mathematical constant
negative_num = -5.5 # Floats can be negative
temprature = 44.5   # temprature
print(price)
print(pi)
print(negative_num)
print(type(temprature))

# output :
#19.99
#3.14159
#-5.5
# class 'float'>
````

# **4. Boolean**

In Python, a boolean (bool) is a data type that has only two possible values: True or False. It acts like a digital light switch that is either ON or OFF to help your program make decisions.

on otherhand ,

In Python, a Boolean (written as bool) is a data type that can only hold one of two possible values: True or False.

Think of a Boolean like a light switch. It can only be in one of two states: either the light is ON (True) or the light is OFF (False). There is no middle ground.

# **Key Properties of a Boolean**
* _Strict Capitalization: You must always capitalize the first letter (True and False)._ 
* _Writing true or false in lowercase will cause a name error._
* _The Result of Comparisons: Any time you use comparison operators like greater than (>), less than (<), or equal to (==), Python automatically answers with a Boolean._
* _Acts as 1 and 0: Under the hood, Python treats True like the integer 1 and False like the integer 0._
* _Controls Program Flow: Booleans are the engine behind if statements, deciding whether a block of code should run or be skipped._

````python
# Example of boolean(bool) vaariable
is_logged_in = True   # The user is signed in
is_game_over = False  # The game is still running
is_client_connected = False # the client is not connected yet
print(is_logged_in)
print(is_game_over)
print(is_client_connected)
print(type(is_client_connected))

# output :
#True
#False
#False
#<class 'bool'>
````

# **5. List**

In Python, a list is simply an ordered collection of items stored inside square brackets [] and separated by commas. It acts like a digital shopping list or a container that lets you keep multiple pieces of data together in one place.

on anotherhand,
In Python, a list is an ordered, changeable collection used to store multiple items in a single variable.

Think of it exactly like a shopping list or a to-do list. You can write items down in a specific order, add new items to the bottom, scratch out items you do not want, or change an item at any time.

# **Key Properties of a List**
* _Uses Square Brackets []: Items are written inside square brackets [] and separated by commas._
* _Ordered and Indexed: Every item has a fixed position starting from index 0._
* _The first item is at [0], the second is at [1], and so on._
* _Changeable (Mutable): You can freely add, remove, or modify items after the list is created._
* _Allows Duplicates and Mixed Data: A single list can hold identical items and combine different data types (like strings, integers, and floats) together._

````python
# Example of list(list) vaariable
fruits = ["apple", "banana", "cherry"]  # A list of strings
numbers = [1, 2, 3, 4, 5]               # A list of integers
mixed = ["Alice", 25, True, 4.5]        # A list with different data types
print(fruits)
print(numbers)
print(mixed)
print(type(fruits))
# output :
#['apple', 'banana', 'cherry']
#[1, 2, 3, 4, 5]
#['Alice', 25, True, 4.5]
#<class 'list'>
````

# **6. Tuple**

In Python, a tuple is an ordered collection of items stored inside parentheses () and separated by commas. It is almost identical to a list, with one major difference: once you create a tuple, it cannot be changed.
in other words , A tuple is a built-in Python data collection used to store multiple items in a single variable. Think of it like a locked box containing a list of fixed facts that cannot be changed once created.
It acts like a locked digital vault or a permanent record of data.

###  **Key Properties of a TupleUses Parentheses:**
 * _Items are written inside round brackets () and separated by commas._
 * _Immutable: You cannot add, remove, or change any item after creation._
 * _Ordered: The items stay in a fixed order, starting from index 0._
 * _Holds Mixed Data: A single tuple can hold floats, strings, integers, or boolean_

````python
# Example of tuple variable
coordinates = (40.7128, -74.0060)      # GPS coordinates (Latitude, Longitude)
rgb_color = (255, 0, 128)               # A fixed color code
screen_size = (1920, 1080)              # Screen resolution dimensions
print(coordinates)
print(rgb_color)
print(screen_size)
print(type(coordinates))
# output :
#(40.7128, -74.006)
#(255, 0, 128)
#(1920, 1080)
#<class 'tuple'>
````

# **8. Set**

In Python, a set is an unordered collection of unique items stored inside curly brackets {}. It acts like a digital sorting basket that automatically filters out duplicates, ensuring that every single item inside appears exactly once.
on the other hand,In Python, a set is an unordered collection of unique items.

Think of it like a bag of lottery balls. You can throw items into the bag, but you cannot have two balls with the exact same number, and they do not stay in any specific row or order.

#  **Key Properties of a Set**

* _Uses Curly Brackets {}: Items are written inside curly brackets {} and separated by commas, but they do not have colons like dictionaries._
* _No Duplicates Allowed: Every item in a set must be 100% unique._
* _If you add the same item twice, Python automatically deletes the duplicate._
* _Unordered and Unindexed: Items do not have a fixed position. You cannot access items using index numbers like my_set[0]._
* _Changeable (Mutable): You can freely add or remove items, but the items inside must be immutable types (like strings, integers, or tuples)._

````python
# Example of set variables
# Duplicates are automatically removed
numbers = {1, 2, 2, 3, 3, 3} 
print(numbers) # Output will just be {1, 2, 3}

# A set of unique words
unique_tags = {"python", "coding", "tutorial"}
print(unique_tags) # output : {'python', 'coding', 'tutorial'}

#  The Raw History (A List that allows duplicates)
visitor_log = ["Ali", "Bilal", "usman", "ali", "farhan", "Ali","farhan"]

#  Filtering Unique Visitors (Convert the list into a Set)
unique_visitors = set(visitor_log)

#  See the magic happen
print("Total Page Views (List):", len(visitor_log)) # output : Total Page Views (List): 7
print("Unique Visitors (Set):  ", unique_visitors) # output : Unique Visitors (Set):   {'farhan', 'Bilal', 'Ali', 'usman', 'ali'}

#  Modifying the Set
unique_visitors.add("farhan")  # Adds David because he is new
unique_visitors.add("arman")   # Python completely ignores this because Alex is already there

print("Updated Unique Visitors:", unique_visitors) # output : Updated Unique Visitors: {'arman', 'farhan', 'Bilal', 'Ali', 'usman', 'ali'}
print(type(numbers))  # output : <class 'set'>

````

# **9.  None (None Type)**

In Python, None is a special keyword used to define a null value or no value at all.Think of it like an empty box. The box exists, but there is absolutely nothing inside it. _It does not mean *0* (which is a number)_, and it does not mean _"" (which is an empty string). It simply means_ **"nothing."**

#  **Key Properties of None**
* _It is its own Data Type: The type of None is called NoneType._
* _It evaluates to False: In conditional if statements, None behaves like a False value._
* _Capital "N" is mandatory: Python is case-sensitive. You must write it as None._
* _Writing none will cause an error._
* _Default Return Value: If a function finishes running but does not explicitly use the return keyword, Python automatically returns None._

````python
# 1. We create a variable for a high score, but no one has played yet
high_score = None

# 2. Let's see what is inside and what data type Python calls it
print("Current Score:")
print(high_score)
print(type(high_score))

print("-" * 20) # Just a dividing line for the output

# 3. Someone plays the game and scores 95 points
high_score = 95

# 4. Let's look inside the box again
print("New Score:")
print(high_score)
print(type(high_score))


# output :
#Current Score:
#None
#<class 'NoneType'>
#--------------------
#New Score:
#95
#<class 'int'>
````

# **10. Complex** 

In Python, a complex data type is used to represent complex numbers. In math, these are numbers that have two parts: a real part (a normal number) and an imaginary part (a number multiplied by square root of -1  ,like this   , which Python represents with a j).Think of it as a 2-in-1 number package used mostly by engineers and scientists to track two related values at the same time (like electrical currents or sound waves).

on the other hand,In Python, complex is a built-in numeric data type used to represent **complex numbers**.Think of it as a number with two separate parts bound together: a real part that you use in everyday math, and an imaginary part used for advanced calculations.

**In mathematics,** _imaginary numbers use an i_, but **in Python,** _the letter j is strictly used._

#  **Key Properties of Complex Numbers**
* _The Mandatory j Suffix: The imaginary part must end with a lowercase j or uppercase J._ 
* _Using i or any other letter will cause a syntax error._
* _No Spaces Allowed: When writing a complex number directly, you cannot put a space between the number and the j (e.g., 3 + 4 j is invalid; it must be 3 + 4j)._
* _Always Stored as Floats: Python automatically stores both the real and imaginary components as floating-point numbers, even if you input them as integers._
* _Immutable: Just like standard integers and floats, complex numbers cannot be modified after they are created._

````python
# 1. Create a complex number (3 is the real part, 5 is the imaginary part)
# Note: You MUST use 'j' in Python, not 'i'
my_number = 3 + 5j

# 2. Print the number and look at its official data type
print("The whole number package:")
print(my_number)
print(type(my_number))

print("-" * 30) # A dividing line

# 3. Pull out the real and imaginary pieces separately
print("The Real part:")
print(my_number.real)

print("The Imaginary part:")
print(my_number.imag)
# Method 1: Direct assignment (Real part = 3, Imaginary part = 4)
num1 = 3 + 4j

# Method 2: Using the complex() function
num2 = complex(5, -2)  # Generates: (5-2j)

print((num1))  


# output :
#The whole number package:
#(3+5j)
#<class 'complex'>
#------------------------------
#The Real part:
#3.0
#The Imaginary part:
#5.0
````

# **Typecasting in python**

## **what is typecasting in python**
### **Typecasting** (ya type conversion) ka matlab hai ek data type ko doosre data type mein badalna (**convert** karna).

Jab aap ek type ke data ko (jaise text ko) doosre type (jaise number) mein change karna chahte hain, toh aap typecasting ka istemal karte hain.

### Imagine kijiye aap user se uski age input lete hain. Python hamesha input ko text (string) samajhta hai. Agar aapko us par math karni hai, toh aapko use whole number (integer) mein badalna padega.

# **The 2 Roles of Type CastingType** 
casting plays two**distinct roles** in Python based on how the conversion happens:
1. Implicit Type Casting (Automatic Conversion)The Role: Python automatically converts one data type to another to prevent data loss, without requiring any action from the programmer.
 How it works: If you add an integer to a float, Python automatically changes the integer into a float behind the scenes.

2. Explicit Type Casting (Manual Conversion)The Role: The programmer forces a variable to change to a specific type using built-in Python constructor functions.
How it works: You use functions like int(), float(), str(), list(), or tuple(). 


# **Core Properties & Rules of Type Casting**
To use type casting safely, you must understand the rules that dictate how data behaves during a conversion:
1. Data Integrity and Loss PropertyWhen casting down from a highly precise type (like a float) to a simpler type (like an integer), Python throws away the decimal part completely. It does not round up; it simply cuts off the tail.
   
```
gpa = 3.99
print(int(gpa))  # Outputs: 3 (Data lost! It does not round to 4)

```

2. Value Compatibility PropertyYou can only explicitly cast data if the original value makes sense in the target format. Trying to cast incompatible text will crash your program with a ValueError.
   
```
valid_cast = int("123")    # Works perfectly (Returns integer 123)
invalid_cast = int("abc")  # Crashes the code! (Python doesn't know how to turn "abc" into a number)

```
3. The Boolean (Truthy/Falsy) Casting PropertyAny data type can be cast into a Boolean (True or False).False values: 0, 0.0, None, empty strings "", empty lists [], empty tuples (), and empty dictionaries {} cast to False.True values: Any number other than zero, or any collection/string that contains at least one item, casts to True.
   
```
print(bool(0))        # Outputs: False
print(bool(55))       # Outputs: True
print(bool("Hello"))  # Outputs: True
print(bool(""))       # Outputs: False

```

4. Complex Number Constraint PropertyWhile you can cast integers and floats into complex numbers, you cannot cast a complex number back into an integer or a float. Doing so will raise a TypeError.

```
# Valid casting into complex:
print(complex(10))  # Outputs: (10+0j)

# Invalid casting out of complex:
num = 5 + 2j
# print(float(num))  # This will cause a TypeError!

```

````python
# 1. Ek string (text) variable banate hain
age_as_text = "25"
print("Pehle data kya hai:", age_as_text)
print("Pehle type kya tha:", type(age_as_text))

print("-" * 30) # Dividing line

# 2. TYPECASTING: int() ka use kar ke text ko number banaya
age_as_number = int(age_as_text)

# 3. Ab result dekhein
print("Typecasting ke baad data:", age_as_number)
print("Ab naya type kya hai:", type(age_as_number))

# output :
# Pehle data kya hai: 25
#Pehle type kya tha: <class 'str'>
#------------------------------
#Typecasting ke baad data: 25
#Ab naya type kya hai: <class 'int'>
````
